```python
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # nameとageを出力するメソッド
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)


# Humanクラスのインスタンスを作成し、変数に代入
human = Human("山田太郎", 20)

# printinfoメソッドを実行
human.printinfo()
```
