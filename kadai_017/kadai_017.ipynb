```python
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 20歳以上かどうかを判定するメソッド
    def check_adult(self):
        if self.age >= 20:
            print(self.name, "は大人です")
        else:
            print(self.name, "は大人ではありません")


# Humanクラスのインスタンスを作成してリストに追加
humans = [
    Human("山田太郎", 19),
    Human("佐藤花子", 20),
    Human("鈴木一郎", 21)
]

# リストの要素数分だけcheck_adultメソッドを呼び出す
for human in humans:
    human.check_adult()
```
